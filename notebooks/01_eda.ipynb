# 01. Exploratory Data Analysis (IEEE-CIS Fraud Detection)

This notebook inspects class balance, missingness, feature distributions, categoricals, and temporal stability.

```python

import pandas as pd
import seaborn as sns

%matplotlib inline
sns.set_theme(style='whitegrid')
```

```python
# Load ingested parquet dataset
df = pd.read_parquet('../data/interim/train_merged.parquet')
print(f'Merged Dataset Shape: {df.shape}')
print(f'Fraud Rate: {df["isFraud"].mean():.4%}')
```

## 1. Class Balance and Time Trends

```python
df['day_index'] = (df['TransactionDT'] // 86400).astype(int)
df['week_index'] = (df['day_index'] // 7).astype(int)
weekly_stats = df.groupby('week_index')['isFraud'].agg(['count', 'mean']).reset_index()
print(weekly_stats.head())
```

## 2. Missingness Analysis

```python
missing_pct = df.isna().mean()
high_missing = missing_pct[missing_pct > 0.90]
print(f'Columns with >90% missingness: {len(high_missing)}')
print(high_missing.head(10))
```

## 3. High Cardinality Categoricals

```python
cat_cols = df.select_dtypes(include=['category', 'object']).columns
for col in ['card1', 'card2', 'addr1', 'P_emaildomain', 'DeviceInfo']:
    if col in df.columns:
        print(f'{col}: {df[col].nunique()} unique categories')
```
